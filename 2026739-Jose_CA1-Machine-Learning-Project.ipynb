```python
import pandas as pd
import numpy as np

# 1. Define column names according to dataset documentation (wdbc.names)
# Column 0: ID, Column 1: Diagnosis (M = Malignant, B = Benign)
# Columns 2 to 31: 30 cellular characteristics (10 features measured in 3 variants: mean, se, worst)
feature_names = [
    'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'smoothness_mean',
    'compactness_mean', 'concavity_mean', 'concave_points_mean', 'symmetry_mean', 'fractal_dimension_mean',
    'radius_se', 'texture_se', 'perimeter_se', 'area_se', 'smoothness_se',
    'compactness_se', 'concavity_se', 'concave_points_se', 'symmetry_se', 'fractal_dimension_se',
    'radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst', 'smoothness_worst',
    'compactness_worst', 'concavity_worst', 'concave_points_worst', 'symmetry_worst', 'fractal_dimension_worst'
]

columns = ['id', 'diagnosis'] + feature_names

# 2. Load the .data file (ensure it is located in the same working directory or provide the correct path)
df = pd.read_csv('wdbc.data', header=None, names=columns)

# 3. Initial exploratory glance
print("Dataset dimensions (Rows, Columns):", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# 4. Check for null values and data types
print("\nGeneral info and missing values:")
print(df.info())

# 5. Distribution of the target variable (Diagnosis: M vs B)
print("\nClass distribution:")
print(df['diagnosis'].value_counts())
```
